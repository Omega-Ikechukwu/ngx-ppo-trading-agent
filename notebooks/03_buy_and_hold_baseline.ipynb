```python
import pandas as pd

df = pd.read_csv("../data/NGX_Stocks_2000-2025.csv")
df.head()
```

```python
# Data filtered for the period of one year to match the PPO agent
df['Date'] = pd.to_datetime(df['Date'])
df_2022 = df[df['Date'].dt.year == 2022].copy()
display(df_2022.head())
```

```python
# Selecting 50 companies
import numpy as np

unique_symbols = df_2022['Symbol'].unique()
selected_symbols = np.random.choice(unique_symbols, size=50, replace=False)
print(selected_symbols)
```

```python
# To determine the stock price at the beginning and end of this period
price_data = {}

for symbol in selected_symbols:
    company_data = df_2022[df_2022['Symbol'] == symbol].copy()
    if not company_data.empty:
        initial_price = company_data.sort_values(by='Date').iloc[0]['Price']
        final_price = company_data.sort_values(by='Date').iloc[-1]['Price']
        price_data[symbol] = {'initial_price': initial_price, 'final_price': final_price}

print(price_data)
```

```python
# Calculating the number of shares bought with 200 Naira at the beginning and end of the time period
investment_results = {}

for symbol, prices in price_data.items():
    initial_price = prices['initial_price']
    final_price = prices['final_price']
    investment_amount = 200

    if initial_price > 0:
        shares_purchased = investment_amount / initial_price
        final_investment_value = shares_purchased * final_price
    else:
        shares_purchased = 0
        final_investment_value = 0

    investment_results[symbol] = {
        'shares_purchased': shares_purchased,
        'final_investment_value': final_investment_value
    }

display(investment_results)
```

```python
# Calculating Profit/Loss
for symbol, results in investment_results.items():
    final_investment_value = results['final_investment_value']
    investment_amount = 200
    profit_loss = final_investment_value - investment_amount
    investment_results[symbol]['profit_loss'] = profit_loss

display(investment_results)
```

```python
# Results summary
investment_amount = 200

for symbol, results in investment_results.items():
    final_investment_value = results['final_investment_value']
    profit_loss = results['profit_loss']
    print(f"Company: {symbol}")
    print(f"  Initial Investment: {investment_amount:.2f} Naira")
    print(f"  Final Value: {final_investment_value:.2f} Naira")
    print(f"  Profit/Loss: {profit_loss:.2f} Naira")
    print("-" * 20)
```

```python
total_final_investment_value = sum(results['final_investment_value'] for results in investment_results.values())

print(f"Total initial investment: 10000.00 Naira")
print(f"Total final investment value: {total_final_investment_value:.2f} Naira")
```

```python
# Initial investment value and Total Investment value
total_initial_investment = 10000
total_profit_loss = total_final_investment_value - total_initial_investment

print(f"Overall Buy and Hold Strategy Performance (2022)")
print(f"Total Initial Investment: {total_initial_investment:.2f} Naira")
print(f"Total Final Investment Value: {total_final_investment_value:.2f} Naira")
print(f"Total Profit/Loss: {total_profit_loss:.2f} Naira")
```
