```python
import pandas as pd

df = pd.read_csv("../data/NGX_Stocks_2000-2025.csv")
df.head()
```

```python
display(df.head())
display(df.info())
display(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

missing_values = df.isnull().sum()
missing_values = missing_values[missing_values > 0]

if not missing_values.empty:
    plt.figure(figsize=(10, 6))
    sns.barplot(x=missing_values.index, y=missing_values.values)
    plt.title('Number of Missing Values per Column')
    plt.xlabel('Columns')
    plt.ylabel('Number of Missing Values')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
else:
    print("No missing values found in the DataFrame.")
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values(by='Date')
```

```python
# Trading Volume over time

plt.figure(figsize=(14, 7))
plt.plot(df['Date'], df['Volume'])
plt.title('Trading Volume Over Time')
plt.xlabel('Date')
plt.ylabel('Volume')
plt.grid(True)
plt.show()
```

```python
# Stock price trend for top 10 sectors
sector_daily_return = df.groupby('Sector')['Daily Return'].mean().sort_values(ascending=False)
top_sectors = sector_daily_return.head(10).index

print("Top 10 Sectors by Average Daily Return:")
print(sector_daily_return.head(10))

plt.figure(figsize=(16, 8))
for sector in top_sectors:
    sector_df = df[df['Sector'] == sector]
    plt.plot(sector_df['Date'], sector_df['Price'], label=sector)

plt.title('Stock Price Trend for Top 10 Sectors')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Market cap trend for top 10 companies
company_market_cap = df.groupby('Company Name')['Market Cap'].mean().sort_values(ascending=False)
top_companies = company_market_cap.head(10).index

print("\nTop 10 Companies by Average Market Cap:")
print(company_market_cap.head(10))

plt.figure(figsize=(16, 8))
for company in top_companies:
    company_df = df[df['Company Name'] == company]
    plt.plot(company_df['Date'], company_df['Market Cap'], label=company)

plt.title('Market Capitalization Trend for Top 10 Companies')
plt.xlabel('Date')
plt.ylabel('Market Cap')
plt.legend()
plt.grid(True)
plt.show()
```
